### **TILES EXPORT UNDERSTAND APPLE MAPS URL**

```python
import numpy as np
import json

center_lat = 21.163879
center_lon = 72.787002
total_lat_span = 0.011456
total_lon_span = 0.020968

tile_lat_span = 0.0027
tile_lon_span = 0.0029

top_lat = center_lat + total_lat_span/2
bottom_lat = center_lat - total_lat_span/2
west_lon = center_lon - total_lon_span/2
east_lon = center_lon + total_lon_span/2

lat_positions = np.arange(top_lat, bottom_lat, -tile_lat_span)
lon_positions = np.arange(west_lon, east_lon, tile_lon_span)

tile_registry = {}  # this is your saved metadata -- keep this dict!
tile_num = 1

for lat in lat_positions:
    for lon in lon_positions:
        tile_center_lat = lat - tile_lat_span/2
        tile_center_lon = lon + tile_lon_span/2

        tile_top_left = (tile_center_lat + tile_lat_span/2, tile_center_lon - tile_lon_span/2)
        tile_bottom_right = (tile_center_lat - tile_lat_span/2, tile_center_lon + tile_lon_span/2)

        tile_id = f"tile_{tile_num:02d}"
        url = f"https://maps.apple.com/frame?map=satellite&center={tile_center_lat:.6f},{tile_center_lon:.6f}&span={tile_lat_span},{tile_lon_span}"

        tile_registry[tile_id] = {
            "url": url,
            "top_left": tile_top_left,
            "bottom_right": tile_bottom_right
        }
        print(f"{tile_id}: {url}")
        tile_num += 1

# Save ALL candidate tiles' metadata now -- you'll delete the ones you don't use later
with open('tile_registry_full.json', 'w') as f:
    json.dump(tile_registry, f, indent=2)

print(f"\n{len(tile_registry)} candidate tiles generated. Pick your favorite 5-6 by their tile_id.")
```
